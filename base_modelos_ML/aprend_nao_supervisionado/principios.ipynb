**Aprendizado Superviosionado X Não Supervisionado**

Supervisionado:     [ Dados (X) ] + [ Rótulos (Y) ]  ---> Modelo aprende a prever Y

Não Supervisionado: [ Dados (X) ]                   ---> Modelo descobre a estrutura

**Objetivo do Aprendizado Não Supervisionado**

Encontrar padrões ocultos dentro dos próprios dados, mas sem um gabarito(rótulo)

Os modelos aprendezem sozinhos a descobrir coisas utilizando algorítimos\
para diversas situações:

- Simplicar conjuntos de dados gigantes e acelerar resultados (Utilizado em visãoc computacional)
- Uso de analogias para detecar padrões, detecção de anomalias
- Realizar agrupamentos

**Conceito Matemático fundamental**
Existe uma distância entre os "pontos de dados" que permite dizer quão parecidos são esses pontos?

Analiticamente existem 2 métodos de cálculode de distancia para responder essas perguntas, e os algoritmos são baseados nesses fundamentos.


- Distancia euclidiana: É basicamente a linha reta entre 2 pontos
- Distancia Manhattan: É a distancia percorrida em 2D ou seja em angulos de 90º  |__


**Casos de uso**

Euclidiana: Dados contínuos, variáveis númericas e poucas dimensões(poucas colunas na base) < 10
(peso, altura,  distancias,idade) agurpando  espaços e coordenadas geograficas, segmentacao de clientes, frequencias de som
Mede sempre a menor distancia entre 2 pontos


Manhatan: Dados de alta dimensionalidade(muitas colunas)
Dados com muitos outliers
Mede distancias cujas caminhos são sememlhantes
determinar rotas, processamento de linguagem natural(previsão da proxima palavra)


**Quando Usar**
- Ausencia de dados rotulados (não conheço bem o que tem na base)
- Criar gurps semelhantes em documentos, produtos, cesta compras, clientes
- Centenas de colunas que precisam ser exploradas, mas reduzir a complexidade sem perder o essencial (PCA)
- Identificar anomalias( invasão de sistemas, fraudes bancárias, doenças)

```python
import numpy as np
from scipy.spatial.distance import euclidean, cityblock

# Cliente A (Padrão): [Idade: 30, Gastos: 100]
cliente_A = np.array([30, 100])

# Cliente B (Semelhante): [Idade: 32, Gastos: 110]
cliente_B = np.array([32, 110])

# Cliente C (Outlier de Gastos): [Idade: 30, Gastos: 1000]
cliente_C = np.array([30, 1000])

print("--- Comparação Cliente A vs Cliente B ---")
print(f"Euclidiana: {euclidean(cliente_A, cliente_B):.2f}")
print(f"Manhattan : {cityblock(cliente_A, cliente_B):.2f}\n")

print("--- Comparação Cliente A vs Cliente C (Outlier) ---")
print(f"Euclidiana: {euclidean(cliente_A, cliente_C):.2f} (Impacto Quadrático!)")
print(f"Manhattan : {cityblock(cliente_A, cliente_C):.2f} (Crescimento Linear)")
```
